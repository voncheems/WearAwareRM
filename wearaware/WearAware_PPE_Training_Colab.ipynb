# WearAware PPE training in Google Colab

This uses Ultralytics Construction-PPE. It includes helmet, gloves, vest, boots, and goggles. It does not include face masks.

```python
!pip -q install -U ultralytics
from ultralytics import YOLO
from google.colab import files
from pathlib import Path
```

The dataset downloads automatically the first time Ultralytics sees `construction-ppe.yaml`. It is released under AGPL-3.0, so keep the required attribution with your demo.

```python
model = YOLO('yolo11n.pt')
results = model.train(
    data='construction-ppe.yaml',
    epochs=80,
    imgsz=640,
    batch=8,
    patience=20,
    project='/content/wearaware-runs',
    name='construction-ppe',
    pretrained=True,
)
```

```python
best = '/content/wearaware-runs/construction-ppe/weights/best.pt'
trained = YOLO(best)
print('Trained classes:', trained.names)
files.download(best)
```

After downloading `best.pt`, back up the current file and replace `services/ppe-api/best.pt`. Restart the PPE API and confirm its startup log lists the new classes. Do not enable mask requirements until the model has also been trained with a mask class.
